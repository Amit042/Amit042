```python
import torch
import torch.nn as nn
import torch.nn.functional as F

print("your torch version is" , torch.__version__)
print("your cuda version is" , torch.cuda.is_available())
if torch.cuda.is_available() == True:
  print("device is gpu suported")
else:
  print("device is cpu supported")

data = """ My name is AMit.
i am try to learn cloud.
hope fully i can full fill my daage.
created by me only one day.
i have seen very much up and down.
i my capacity but still agger.
to do mistake and learn some new from that mistake.
sometime it bit confusion the way .
i take my decisions of life balance.
completly forget what should i pkan.
how to approch to my plan.
right now you can consider me as a fail father.
but it is my promissed on day.
i willl bring back back my all losses."""
data1 = data.splitlines()

#print(data1)

words = []
for line in data1:
  line = line.replace("." , "")
  line1 =  line  +  "  <END>"
  #print(line1)
  words.append(line1)
print(words)

new_data = " ".join(words)
new_data = new_data.split()
#print(new_data)
print(len(new_data))
voceb = sorted(set(new_data))
#print(voceb)
print (len(voceb))


token_to_id = {
    token:i
    for i,token in enumerate(voceb)
}
#print(token_to_id)

token_idx = [token_to_id[token]for token in new_data]
#print(token_idx)

block_size = 8
embedding_dim = 32
embedding = nn.Embedding(len(voceb),embedding_dim)
torch_tensor = torch.tensor(token_idx)
token_block = torch_tensor[:block_size]
token_vector = embedding(token_block)

#print(token_vector)


positions = torch.arange(block_size)
position_embedding = nn.Embedding(block_size, embedding_dim)
position_vector = position_embedding(positions)

#print(position_vector)



```
